# 38 · How far do the R5 biology stories hold in independent segment-resolved data?

Notebooks 31–34 found species-by-position differences in PT gene expression and a literature check
placed a short list of them as candidate novel biology. This notebook asks, story by story,
whether each claim survives three things notebook 34 could not test:

1. a **multi-donor human atlas with true S1/S2/S3 labels, restricted to cortex** (Lake et al. 2023,
   KPMP snRNA v1.5), instead of a convoluted-versus-S3 annotation;
2. **female mice measured directly** (microdissected PT segments, Chen et al. 2023) as well as by
   snRNA;
3. the **probe panels**: genes whose two species panels carry different numbers of probes.

It fits nothing new on our data. Our side is the per-specimen segment means and notebook 12's
fitted curves, reproduced exactly.

| Dataset | Species | What was measured | Sex | Role here |
|---|---|---|---|---|
| Our Visium HD tubules (2 + 2 specimens) | both | reviewed S1/S2/S3 structures | male | the claims |
| Lake et al. 2023, CELLxGENE `a12ccb9b` | human | snRNA, author labels PT-S1/S2/S3, `region` C/C-M/M | both | **primary human reference** |
| CELLxGENE Census `25818bf7` (notebook 34 pseudobulk) | mouse | snRNA, PT segment 1/2/3 | both | mouse reference by sex |
| Chen et al. 2021, GSE150338 | mouse | microdissected PTS1/2/3, bulk RNA-seq | male | mouse reference (anatomical) |
| Chen et al. 2023, GSE212213 | mouse | microdissected PTS1/2/3, bulk RNA-seq, mean TPM | both | female anatomical reference |
| Lee et al. 2015, GSE56743 | rat | microdissected S1/2/3 | male | reported only |
| CELLxGENE Census `09b518f9` (notebook 34) | human | snRNA, convoluted PT vs S3 | both | secondary human reference |
| 10x probe sets; our `molecule_info.h5` | both | probes per gene | — | panel balance |
| Human Protein Atlas | human | kidney IHC | — | protein presence only |

Anatomy (ESBL/MRECA definitions): mouse PTS1 is the initial convoluted tubule, PTS2 the straight
tubule in **cortical medullary rays**, PTS3 the straight segment in the **outer stripe**. Our human
sections are cortex, so mouse PTS2 is the anatomically closest match to human cortical S3.

## 0 · Protocol, fixed before any result in this notebook

**Question.** For each R5 gene, does its species-by-position claim hold when the human comparator
is a cortex-restricted multi-donor atlas with true S1/S2/S3 labels, the mouse comparator includes
females, and probe-panel imbalance is flagged?

**Prior.** The four lead stories (creatine synthesis, mitochondrial fatty-acid oxidation, drug
conjugation versus glutathione placement, the Dcxr reversal) should hold; they already replicate in
four external datasets. The sterol block and the one-species rows are expected to weaken. Either
outcome is reported.

**Contrasts.** Every arm is late − early = **S3 − S1** on that dataset's own log scale (ours:
natural-log normalized; snRNA: log2 CPM computed against all-gene totals; microdissection:
log2 TPM+1). The species-by-position interaction is (human S3 − S1) − (mouse S3 − S1). Signs and
within-dataset ratios are compared, never absolute levels across platforms.

**Human reference.** Lake snRNA v1.5 nuclei with `disease == normal`, author label PT-S1/2/3,
`region == C` and `tissue == cortex of kidney`; donors with at least 50 nuclei in each segment
(primary); at least 20 (sensitivity). Per-donor pseudobulk of raw counts divided by the nuclei's
total counts over all genes, log2 CPM+1, as in notebook 34. A donor-level contrast is averaged
over donors.

**Claims.** The table in the next cell, written from notebooks 32–34 and the phase-1 audit before
this notebook ran. `+`/`−` = sign of S3 − S1; `0` = flat relative to the other species;
blank = not claimed. One-species genes are claimed as expressed in one species only.

**Tests per gene** (each True, False or not testable):
- `ours`: every claimed arm holds in our data, in the species mean **and** in each specimen
  (signed arms) or in all four human–mouse specimen pairings (interaction).
- `ours_attenuation`: a human `0` claim still holds after dividing our human contrast by the
  attenuation factor *a* estimated in section 6 (only when *a* < 1).
- `human_lake`: the human arm holds in the Lake primary reference. For `0`, relative to the mean of
  the male and female mouse snRNA contrasts.
- `int_vs_male_sn`, `int_vs_female_sn`: the interaction sign with Lake minus mouse snRNA by sex.
- `mouse_sn_male`, `mouse_sn_female`, `mouse_micro_male`, `mouse_micro_female`: the mouse arm.
- One-species genes: `mouse_absent` (microdissection maximum segment TPM < 5 in both sexes) for
  human-only genes; `human_absent` (Lake maximum segment CPM < 10) for mouse-only genes.
- Not testable: a reference whose maximum segment value is below the same floors (CPM 10, TPM 5)
  for a gene claimed to be expressed there.
- A signed arm needs |contrast| ≥ 0.1; a `0` arm needs |contrast| ≤ 0.5 × the other species'
  contrast, which must itself be ≥ 0.1 in magnitude.

**Gene verdicts** (`pseudospace.replication_rules.gene_verdict`). Critical tests: `ours`,
`human_lake`, `mouse_absent`, `human_absent`. Any failed critical test or two failed tests →
*not replicated*. No failure, every critical test evaluated, probes 1:1 and detection ≥ 10% of
structures in both species (or a one-species label) → *headline*. Otherwise → *supporting*.

**Story verdicts** (`story_verdict`). A story stays a **lead** if every lead gene is headline and
at least 60% of its probe-balanced members are headline or supporting; with the member rule alone
it is supporting. Stories already demoted in phase 1 can at most be supporting.

**Attenuation.** Workstream 1 is testing whether noisier human positions could flatten human
curves. A segment-label version is estimated here: among control genes strongly zonated in the
same direction in both external species (|S3 − S1| ≥ 1 in Lake and in male mouse snRNA,
detected in ≥ 10% of our structures in both species, probes 1:1), *a* = median(our human / Lake)
÷ median(our mouse / mouse snRNA). It is chosen by external data only.

**Human region (descriptive).** For donors with ≥ 30 nuclei in cortical S1, cortical S3 and
medullary S3 (`region` M or C-M), human medullary S3 − cortical S1 is compared with the mouse
contrast. A human `0` claim is flagged *region-dependent in human* if human medullary S3 moves at
least half of the mouse contrast in the mouse direction.

**What would make me abandon a story.** Creatine: GATM falls from human S1 to S3 in the Lake
cortex reference. Fatty-acid oxidation: ACADM and ACAA2 move in the mouse direction there.
Glutathione: female mouse microdissection shows no S1 → S3 rise. Dcxr: DCXR does not rise from
human S1 to S3.

## 1 · Our side: per-specimen segment means and notebook 12's fitted curves

Same PT structures, coordinate support, normalization and 2% detection rule as notebooks 12 and 34.
Segment means use the reviewed S1/S2/S3 labels, not the coordinate. The fitted curves are refit
for the claimed genes only; each gene's fit is independent, and the statistics are checked against
notebook 12's saved values.

```python
import anndata as ad
import numpy as np
from scipy import sparse
from IPython.display import display
from pseudospace.pathway_inputs import rebuild_pt_expression
from pseudospace.pathway_remodeling import fit_nested_trajectories, gene_covariates
from pseudospace.replication_rules import arm_passes, csr_group_sums, gene_verdict, story_verdict
from pseudospace.stage_cache import cached_payload, digest
import pseudospace.pathway_remodeling as remodeling
import pseudospace.replication_rules as rules

saved_pt = ad.read_h5ad(upstream13 / 'cross_species_pt_scfates.h5ad', backed='r')
pt_obs = saved_pt.obs.copy()
saved_pt.file.close()
pt_obs = pt_obs[pt_obs.broad_tubule_marker_call.astype(str).eq('PT')].copy()
expected = {'mouse': ['Ctrl1A2', 'Ctrl1A4'], 'human': ['HUK1_COR1', 'HUK1_MED1']}
support = pt_obs.groupby('sample', observed=True).shared_pseudospace.quantile([.01, .99]).unstack()
coordinate = pt_obs.shared_pseudospace.to_numpy(float)
pt_obs = pt_obs[(coordinate >= support[.01].max()) & (coordinate <= support[.99].min())].copy()
orthologs = pd.read_csv(upstream13 / 'ortholog_map_used.csv')
adata = rebuild_pt_expression(pt_obs, {n: data_root / 'tubule_by_gene' / f'{n}_tubule_by_gene_caleb.h5ad'
                                       for v in expected.values() for n in v}, orthologs, target_sum=1e4)
position = adata.obs.shared_pseudospace.to_numpy(float)
human = adata.obs.comparison_species.astype(str).eq('human').to_numpy()
specimen = adata.obs['sample'].astype(str).to_numpy()
segment = adata.obs.segment_class.astype(str).str.replace('PT-', '', regex=False).to_numpy()
expression = sparse.csr_matrix(adata.layers['lognorm'], dtype=float)
universe = gene_covariates(expression, position, human, specimen, adata.var_names)
universe['eligible'] = adata.var.measured_in_both_inputs.to_numpy() & universe.detection.ge(.02)
genes = pd.Index(universe.loc[universe.eligible, 'gene'])
Y = expression[:, universe.eligible.to_numpy()]
detection = universe.set_index('gene').loc[genes, ['detection_mouse', 'detection_human']]
missing = sorted(set(CLAIMS.gene) - set(genes))
if missing:
    raise ValueError(f'Claimed genes outside the tested universe: {missing}')

names = expected['mouse'] + expected['human']
ours = {}
for name in names:
    for seg in ('S1', 'S2', 'S3'):
        ours[(name, seg)] = np.asarray(Y[(specimen == name) & (segment == seg)].mean(axis=0)).ravel()
ours = pd.DataFrame(ours, index=genes)
ours_contrast = pd.DataFrame({name: ours[(name, 'S3')] - ours[(name, 'S1')] for name in names})
ours_h = ours_contrast[expected['human']].mean(axis=1)
ours_m = ours_contrast[expected['mouse']].mean(axis=1)
ours_int = ours_h - ours_m

story_genes = CLAIMS.gene.tolist()
columns = genes.get_indexer(story_genes)
lo = max(position[specimen == n].min() for n in names)
hi = min(position[specimen == n].max() for n in names)
grid = np.linspace(lo, hi, 61)
fit = cached_payload('story_gene_curves',
    lambda: dict(fit_nested_trajectories(Y[:, columns], position, human, specimen, grid, basis_df=6)),
    root=output / 'stage_cache', params={'basis_df': 6, 'grid': grid, 'genes': story_genes},
    inputs={'Y': digest(Y[:, columns]), 'position': position, 'human': human, 'specimen': specimen},
    code=digest([NOTEBOOK_LOGIC_VERSION, Path(remodeling.__file__)]))
saved_stats = pd.read_csv(upstream12 / 'gene_statistics.csv', index_col=0)
np.testing.assert_allclose(fit['T_spatial'], saved_stats.loc[story_genes, 'T_spatial'], rtol=1e-6, atol=1e-8)
human_curve = pd.DataFrame(fit['human'], index=story_genes, columns=grid)
mouse_curve = pd.DataFrame(fit['mouse'], index=story_genes, columns=grid)
print(f'{len(position):,} structures · {len(genes):,} genes · story-gene curves reproduce notebook 12 T_spatial.')
```

## 2 · Probe panels: probes per gene in each species

Mouse: the public 10x mouse v2.0 panel, whose included genes equal our mouse matrices' genes.
Human: the panel recorded in our own run (`molecule_info.h5`, both sections). A probe counts when
the panel marks it as included. The table covers every accepted ortholog pair; workstream 2 uses it
as a detection-asymmetry filter.

```python
import h5py

mouse_panel = pd.read_csv(MOUSE_PROBES, comment='#')
mouse_panel = mouse_panel[mouse_panel.included.astype(str).str.upper().eq('TRUE')]
mouse_count = mouse_panel.groupby('gene_id').size()
human_counts = []
for section in ('HUK_COR_1', 'HUK_MED_1'):
    with h5py.File(data_root / section / 'outs' / 'molecule_info.h5', 'r') as handle:
        probes = handle['probes']
        table = pd.DataFrame({'feature_id': probes['feature_id'][:].astype(str),
                              'included': probes['included_probes'][:].astype(bool)})
    human_counts.append(table[table.included].groupby('feature_id').size())
pd.testing.assert_series_equal(human_counts[0], human_counts[1])
human_count = human_counts[0]
public_human = pd.read_csv(HUMAN_PROBES, comment='#')
public_human = public_human[public_human.included.astype(str).str.upper().eq('TRUE')].groupby('gene_id').size()

mouse_ids = ad.read_h5ad(data_root / 'tubule_by_gene' / 'Ctrl1A2_tubule_by_gene_caleb.h5ad', backed='r')
mouse_var = mouse_ids.var[['gene_ids']].copy()
mouse_ids.file.close()
human_ids = ad.read_h5ad(data_root / 'tubule_by_gene' / 'HUK1_COR1_tubule_by_gene_caleb.h5ad', backed='r')
human_var = human_ids.var[['gene_ids']].copy()
human_ids.file.close()
if set(mouse_var.gene_ids) != set(mouse_count.index):
    raise ValueError('Mouse matrices and the mouse v2.0 panel disagree; check which panel was used.')

probe_table = orthologs[['mouse_symbol', 'human_symbol']].copy()
probe_table['mouse_gene_id'] = probe_table.mouse_symbol.map(mouse_var.gene_ids)
probe_table['human_gene_id'] = probe_table.human_symbol.map(human_var.gene_ids)
probe_table['mouse_probes'] = probe_table.mouse_gene_id.map(mouse_count).fillna(0).astype(int)
probe_table['human_probes'] = probe_table.human_gene_id.map(human_count).fillna(0).astype(int)
probe_table['human_probes_public_v2'] = probe_table.human_gene_id.map(public_human).fillna(0).astype(int)
lo_p = probe_table[['mouse_probes', 'human_probes']].min(axis=1)
hi_p = probe_table[['mouse_probes', 'human_probes']].max(axis=1)
probe_table['probe_balance'] = np.select(
    [probe_table.mouse_probes.eq(probe_table.human_probes),
     probe_table.mouse_probes.lt(probe_table.human_probes)], ['balanced', 'mouse_fewer'], 'human_fewer')
probe_table['single_vs_multi'] = lo_p.eq(1) & hi_p.ge(2)
probe_table['in_pt_universe'] = probe_table.mouse_symbol.isin(genes)
probe_table = probe_table.merge(detection, left_on='mouse_symbol', right_index=True, how='left')
probe_table.to_csv(output / 'probe_counts.csv', index=False)
(output / 'probe_counts_SOURCE.txt').write_text(
    'probe_counts.csv: included Visium HD probes per gene for every accepted human-mouse ortholog pair.\n'
    'mouse_probes: 10x Visium Mouse Transcriptome Probe Set v2.0 (mm10-2020-A), included == TRUE, counted per Ensembl gene id;\n'
    '  its included genes equal the genes of our mouse tubule_by_gene matrices.\n'
    'human_probes: the probe set recorded in our own Space Ranger run (data/HUK_COR_1 and HUK_MED_1 molecule_info.h5,\n'
    '  identical in both sections), included_probes == True, counted per feature id. human_probes_public_v2: the public\n'
    '  10x Human v2.0 (GRCh38-2020-A) file, for cross-checking only.\n'
    'probe_balance: balanced / mouse_fewer / human_fewer; single_vs_multi: one species has 1 probe and the other >= 2.\n'
    'detection_mouse / detection_human: fraction of our PT structures with the gene detected (notebook 12 universe only).\n'
    f'Written by notebook 38, logic version {NOTEBOOK_LOGIC_VERSION}.\n')
universe_probes = probe_table[probe_table.in_pt_universe]
print(f'{len(probe_table):,} ortholog pairs; in the PT universe ({len(universe_probes):,} genes): '
      f'{universe_probes.probe_balance.eq("balanced").mean():.1%} balanced, '
      f'{universe_probes.single_vs_multi.sum()} single-versus-multi '
      f'({(universe_probes.single_vs_multi & universe_probes.mouse_probes.eq(1)).sum()} with one mouse probe).')
display(probe_table.set_index('mouse_symbol').loc[story_genes, ['human_symbol', 'mouse_probes', 'human_probes', 'probe_balance']].T)
```

## 3 · Human reference: Lake et al. 2023 cortex S1/S2/S3, per donor

Raw counts are summed per donor × segment × region without loading the whole matrix. Region is
*cortex* when `region == C` and `tissue == cortex of kidney`, *medulla* when `region` is M or C-M.
Genes are matched to our ortholog map by Ensembl id (symbol as fallback); duplicated targets are
summed.

```python
def read_obs_column(group, name):
    item = group[name]
    if isinstance(item, h5py.Group):
        categories = np.array([c.decode() if isinstance(c, bytes) else str(c) for c in item['categories'][:]])
        return pd.Categorical.from_codes(item['codes'][:], categories).astype(str)
    values = item[:]
    return values.astype(str) if values.dtype.kind in 'SO' else values


def lake_pseudobulk():
    with h5py.File(LAKE, 'r') as handle:
        obs = pd.DataFrame({c: read_obs_column(handle['obs'], c)
                            for c in ['subclass.l2', 'disease', 'region', 'tissue', 'sex', 'donor_id']})
        var_ids = handle['raw/var/_index'][:].astype(str)
        var_names = np.asarray(read_obs_column(handle['raw/var'], 'feature_name'))
        keep = obs['subclass.l2'].isin(['PT-S1', 'PT-S2', 'PT-S3']) & obs.disease.eq('normal')
        region = np.where(obs.region.eq('C') & obs.tissue.eq('cortex of kidney'), 'cortex',
                          np.where(obs.region.isin(['M', 'C-M']), 'medulla', ''))
        keep &= region != ''
        label = (obs.donor_id + '|' + obs.sex + '|' + obs['subclass.l2'].str.replace('PT-', '', regex=False)
                 + '|' + pd.Series(region, index=obs.index))[keep]
        groups, codes = np.unique(label.to_numpy(), return_inverse=True)
        raw = handle['raw/X']
        sums, totals, counts = csr_group_sums(raw['data'], raw['indices'], raw['indptr'],
                                              np.flatnonzero(keep.to_numpy()), codes, len(groups), len(var_ids),
                                              block_rows=5000)
    # Ensembl first, then symbol, onto our mouse gene names.
    by_id = dict(zip(orthologs.human_symbol.map(human_var.gene_ids), orthologs.mouse_symbol))
    by_symbol = dict(zip(orthologs.human_symbol, orthologs.mouse_symbol))
    target = np.array([by_id.get(i, by_symbol.get(s, '')) for i, s in zip(var_ids, var_names)])
    mapped = target != ''
    frame = pd.DataFrame(sums[:, mapped], columns=target[mapped]).T.groupby(level=0).sum().T
    frame = frame.reindex(columns=genes, fill_value=0.0)
    return {'groups': groups.astype(str), 'counts': counts, 'totals': totals, 'sums': frame.to_numpy(),
            'genes': np.asarray(genes, dtype=str), 'mapped_lake_genes': np.array([mapped.sum()])}


lake = cached_payload('lake_pt_pseudobulk', lake_pseudobulk, root=output / 'stage_cache',
                      params={'regions': 'C/cortex; M,C-M/medulla', 'labels': 'PT-S1,PT-S2,PT-S3', 'disease': 'normal'},
                      inputs={'lake_file': [LAKE.name, LAKE.stat().st_size], 'genes': list(genes),
                              'orthologs': digest(orthologs)},
                      code=digest([NOTEBOOK_LOGIC_VERSION, Path(rules.__file__)]))
lake_meta = pd.DataFrame([g.split('|') for g in lake['groups']], columns=['donor', 'sex', 'segment', 'region'])
lake_meta['nuclei'], lake_meta['total'] = lake['counts'], lake['totals']
lake_cpm = pd.DataFrame(lake['sums'] / lake['totals'][:, None] * 1e6, columns=genes)
lake_log = np.log2(lake_cpm + 1)


def lake_donors(min_nuclei, segments=('S1', 'S2', 'S3'), region='cortex'):
    block = lake_meta[lake_meta.region.eq(region)].pivot_table(index=['donor', 'sex'], columns='segment',
                                                                 values='nuclei', aggfunc='sum', fill_value=0)
    block = block.reindex(columns=list(segments), fill_value=0)
    return block[(block >= min_nuclei).all(axis=1)].reset_index()


def lake_values(donors, seg, region='cortex'):
    rows = [lake_meta.index[(lake_meta.donor == d) & (lake_meta.segment == seg) & (lake_meta.region == region)][0]
            for d in donors]
    return lake_log.loc[rows].set_axis(list(donors))


def lake_contrast(donors, late='S3', early='S1', late_region='cortex'):
    return (lake_values(donors, late, late_region) - lake_values(donors, early)).mean(axis=0)


primary = lake_donors(MIN_DONOR_NUCLEI)
sensitivity = lake_donors(MIN_DONOR_NUCLEI_SENSITIVITY)
print(f'Lake genes mapped to our ortholog map: {int(lake["mapped_lake_genes"][0]):,}')
print('Primary donors:', len(primary), primary.groupby('sex').size().to_dict(),
      '· sensitivity donors:', len(sensitivity), sensitivity.groupby('sex').size().to_dict())
display(primary)
lake_h = lake_contrast(primary.donor)
lake_h_s2 = lake_contrast(primary.donor, late='S2')
lake_h_sens = lake_contrast(sensitivity.donor)
lake_h_male = lake_contrast(primary.donor[primary.sex.eq('male')])
lake_h_female = lake_contrast(primary.donor[primary.sex.eq('female')])
lake_segment_cpm = pd.DataFrame({s: lake_cpm.loc[[lake_meta.index[(lake_meta.donor == d) & (lake_meta.segment == s)
                                                                  & (lake_meta.region == 'cortex')][0]
                                                      for d in primary.donor]].mean() for s in ('S1', 'S2', 'S3')})
lake_segment_log = pd.DataFrame({s: lake_values(primary.donor, s).mean() for s in ('S1', 'S2', 'S3')})
pd.DataFrame({'donor': lake_meta.donor, 'sex': lake_meta.sex, 'segment': lake_meta.segment, 'region': lake_meta.region,
              'nuclei': lake_meta.nuclei}).to_csv(output / 'lake_groups.csv', index=False)
lake_segment_log.add_prefix('lake_log2cpm_').to_csv(output / 'lake_cortex_segment_means.csv')
```

```python
def census(label, segment_map):
    counts = pd.read_csv(external / 'census_pt_segments' / f'{label}_pt_segment_pseudobulk_counts.csv.gz', index_col=0)
    meta = pd.read_csv(external / 'census_pt_segments' / f'{label}_pt_segment_pseudobulk_meta.csv', index_col=0)
    meta = meta[~meta.development_stage.str.contains('embryo|E16|Theiler', case=False)]
    meta['segment'] = meta.cell_type.map(segment_map)
    meta = meta[meta.segment.notna() & meta.n_cells.ge(MIN_CELLS)]
    needed = set(segment_map.values())
    complete = meta.groupby('donor_id').segment.apply(lambda s: needed.issubset(set(s)))
    meta = meta[meta.donor_id.isin(complete[complete].index)]
    cpm = counts.loc[meta.index].div(meta.total_counts_all_genes, axis=0) * 1e6
    cpm = cpm.reindex(columns=genes)
    return np.log2(cpm + 1), cpm, meta


mouse_sn_log, mouse_sn_cpm, mouse_sn_meta = census('mouse', {
    'epithelial cell of proximal tubule segment 1': 'S1', 'epithelial cell of proximal tubule segment 2': 'S2',
    'epithelial cell of proximal tubule segment 3': 'S3'})


def sn_contrast(log, meta, sex, late='S3', early='S1'):
    m = meta[meta.sex.eq(sex)]
    per_donor = []
    for donor, block in m.groupby('donor_id'):
        late_v = log.loc[block.index[block.segment.eq(late)]].mean()
        early_v = log.loc[block.index[block.segment.eq(early)]].mean()
        per_donor.append(late_v - early_v)
    return pd.concat(per_donor, axis=1).mean(axis=1), m.donor_id.nunique()


def sn_segment_means(log, meta, sex):
    m = meta[meta.sex.eq(sex)]
    return pd.DataFrame({s: log.loc[m.index[m.segment.eq(s)]].mean() for s in ('S1', 'S2', 'S3')})


sn_m, n_sn_m = sn_contrast(mouse_sn_log, mouse_sn_meta, 'male')
sn_f, n_sn_f = sn_contrast(mouse_sn_log, mouse_sn_meta, 'female')
sn_seg_m, sn_seg_f = sn_segment_means(mouse_sn_log, mouse_sn_meta, 'male'), sn_segment_means(mouse_sn_log, mouse_sn_meta, 'female')
sn_max_cpm = {sex: pd.concat([mouse_sn_cpm.loc[mouse_sn_meta.index[mouse_sn_meta.sex.eq(sex) & mouse_sn_meta.segment.eq(s)]].mean()
                              for s in ('S1', 'S2', 'S3')], axis=1).max(axis=1) for sex in ('male', 'female')}
print(f'Mouse snRNA donors: {n_sn_m} male, {n_sn_f} female')

male_bulk = pd.read_csv(external / 'segment_microdissection' / 'GSE150338_Mouse_TPM_Replicates.txt.gz', sep='\t')
male_bulk = male_bulk.drop_duplicates('Gene_symbol').set_index('Gene_symbol')
male_tpm = pd.DataFrame({s: male_bulk[[c for c in male_bulk.columns if f'_PT{s}_' in c]].apply(pd.to_numeric, errors='coerce')
                         .clip(lower=0).mean(axis=1) for s in ('S1', 'S2', 'S3')})
sex_bulk = pd.read_csv(external / 'segment_microdissection_sex' / 'GSE212213_TPM_Male_Female.txt.gz', sep='\t')
sex_bulk = sex_bulk.drop_duplicates('Gene_symbol').set_index('Gene_symbol')
female_tpm = sex_bulk[['fPTS1', 'fPTS2', 'fPTS3']].set_axis(['S1', 'S2', 'S3'], axis=1).apply(pd.to_numeric, errors='coerce')
male_tpm_2023 = sex_bulk[['PTS1', 'PTS2', 'PTS3']].set_axis(['S1', 'S2', 'S3'], axis=1).apply(pd.to_numeric, errors='coerce')
rat_bulk = pd.read_csv(external / 'segment_microdissection' / 'GSE56743_RPKM.without.PAseq.txt.gz', sep='\t')
rat_bulk = rat_bulk.drop_duplicates('Symbol').set_index('Symbol')
rat_tpm = pd.DataFrame({s: rat_bulk[[c for c in rat_bulk.columns if f'{s}_' in c]].apply(pd.to_numeric, errors='coerce')
                        .clip(lower=0).mean(axis=1) for s in ('S1', 'S2', 'S3')})
rat_tpm.index = rat_tpm.index.astype(str).str.upper()
rat_tpm = rat_tpm[~rat_tpm.index.duplicated()].reindex(genes.str.upper()).set_axis(genes)
male_tpm, female_tpm, male_tpm_2023 = (t.reindex(genes) for t in (male_tpm, female_tpm, male_tpm_2023))
micro = {name: np.log2(t + 1) for name, t in (('male', male_tpm), ('female', female_tpm), ('rat', rat_tpm))}
micro_contrast = {name: v.S3 - v.S1 for name, v in micro.items()}
micro_s2 = {name: v.S2 - v.S1 for name, v in micro.items()}
check = np.log2(male_tpm_2023 + 1).stack().corr(np.log2(male_tpm + 1).stack(), method='spearman')
print(f'Male microdissection, GSE212213 means versus GSE150338 replicate means: Spearman {check:.3f}')

human_bench = pd.read_csv(external / 'census_pt_segments' / 'human_pt_segment_pseudobulk_counts.csv.gz', index_col=0)
bench_meta = pd.read_csv(external / 'census_pt_segments' / 'human_pt_segment_pseudobulk_meta.csv', index_col=0)
bench_meta['segment'] = bench_meta.cell_type.map({'kidney proximal convoluted tubule epithelial cell': 'early',
                                                  'epithelial cell of proximal tubule segment 3': 'S3'})
bench_meta = bench_meta[bench_meta.segment.notna() & bench_meta.n_cells.ge(MIN_CELLS)]
complete = bench_meta.groupby('donor_id').segment.nunique().eq(2)
bench_meta = bench_meta[bench_meta.donor_id.isin(complete[complete].index)]
bench_log = np.log2(human_bench.loc[bench_meta.index].div(bench_meta.total_counts_all_genes, axis=0) * 1e6 + 1).reindex(columns=genes)
bench_h = (bench_log.loc[bench_meta.segment.eq('S3')].groupby(bench_meta.donor_id[bench_meta.segment.eq('S3')]).mean()
           - bench_log.loc[bench_meta.segment.eq('early')].groupby(bench_meta.donor_id[bench_meta.segment.eq('early')]).mean()).mean()
```

## 6 · Attenuation check from segment labels

If our human segment means were blurred (positions or labels mixed along the axis), every human
contrast would shrink relative to the external references by a common factor, and a "human flat"
claim could be an artefact. Control genes are chosen by the external data alone (protocol).

```python
balanced_genes = set(probe_table.loc[probe_table.probe_balance.eq('balanced'), 'mouse_symbol'])
controls = (detection.min(axis=1).ge(.10) & genes.to_series(index=genes).isin(balanced_genes) & lake_ok
            & lake_h.abs().ge(1) & sn_m.abs().ge(1) & (np.sign(lake_h) == np.sign(sn_m)))
ratio_h = (ours_h / lake_h)[controls]
ratio_m = (ours_m / sn_m)[controls]
attenuation = float(ratio_h.median() / ratio_m.median())
rng = np.random.default_rng(38)
boot = []
index = np.flatnonzero(controls.to_numpy())
for _ in range(2000):
    pick = rng.choice(index, size=len(index), replace=True)
    boot.append(np.median((ours_h / lake_h).iloc[pick]) / np.median((ours_m / sn_m).iloc[pick]))
attenuation_ci = np.quantile(boot, [.025, .975])
attenuation_table = pd.DataFrame({'ours_human': ours_h[controls], 'lake_human': lake_h[controls],
                                  'ours_mouse': ours_m[controls], 'mouse_sn_male': sn_m[controls]})
attenuation_table.to_csv(output / 'attenuation_controls.csv')
print(f'{int(controls.sum())} control genes; median our/external slope: human {ratio_h.median():.3f}, '
      f'mouse {ratio_m.median():.3f}; relative human attenuation a = {attenuation:.2f} '
      f'(gene bootstrap 95% interval {attenuation_ci[0]:.2f}–{attenuation_ci[1]:.2f}).')
print(f'Spearman across controls: human {spearmanr(ours_h[controls], lake_h[controls]).statistic:.2f}, '
      f'mouse {spearmanr(ours_m[controls], sn_m[controls]).statistic:.2f}')
```

## 7 · Apply the pre-specified rules

```python
def pairing_ok(gene, claim):
    if claim in ('', None):
        return None
    values = [ours_contrast.loc[gene, h] - ours_contrast.loc[gene, m] for h in expected['human'] for m in expected['mouse']]
    return all(arm_passes(claim, v) for v in values)


def specimens_ok(gene, claim, species):
    if claim in ('', None, '0'):
        return None
    return all(arm_passes(claim, ours_contrast.loc[gene, n]) for n in expected[species])


def claimed_tests(r):
    """Which tests a claim row asks for (unclaimed tests are omitted, not 'not testable')."""
    shared = r.one_species == ''
    keys = ['ours']
    if r.human == '0' and attenuation < 1:
        keys.append('ours_attenuation')
    if r.human != '':
        keys.append('human_lake')
    if r.interaction != '' and shared:
        keys += ['int_vs_male_sn', 'int_vs_female_sn']
    if r.mouse != '' and r.one_species != 'human':
        keys += ['mouse_sn_male', 'mouse_sn_female', 'mouse_micro_male', 'mouse_micro_female']
    if r.one_species == 'human':
        keys.append('mouse_absent')
    if r.one_species == 'mouse':
        keys.append('human_absent')
    return keys


def max_or_nan(values):
    values = np.asarray(values, float)
    return np.nan if np.isnan(values).all() else np.nanmax(values)


evidence_rows = []
for row in CLAIMS.itertuples():
    g = row.gene
    one = row.one_species
    sn_mean = (sn_m[g] + sn_f[g]) / 2
    human_testable = lake_segment_cpm.loc[g].max() >= CPM_FLOOR
    ours_parts = [arm_passes(row.mouse, ours_m[g], ours_h[g]), arm_passes(row.human, ours_h[g], ours_m[g]),
                  arm_passes(row.interaction, ours_int[g]), specimens_ok(g, row.mouse, 'mouse'),
                  specimens_ok(g, row.human, 'human'), pairing_ok(g, row.interaction)]
    ours_parts = [p for p in ours_parts if p is not None]
    all_tests = {
        'ours': all(ours_parts) if ours_parts else None,
        'ours_attenuation': arm_passes('0', ours_h[g] / attenuation, ours_m[g]),
        'human_lake': arm_passes(row.human, lake_h[g], sn_mean) if human_testable else None,
        'mouse_absent': (None if np.isnan(max_or_nan(male_tpm.loc[g])) or np.isnan(max_or_nan(female_tpm.loc[g]))
                         else bool(max_or_nan(male_tpm.loc[g]) < TPM_FLOOR and max_or_nan(female_tpm.loc[g]) < TPM_FLOOR)),
        'human_absent': bool(lake_segment_cpm.loc[g].max() < CPM_FLOOR),
    }
    for sex, sn in (('male', sn_m), ('female', sn_f)):
        sn_testable = sn_max_cpm[sex][g] >= CPM_FLOOR
        all_tests[f'int_vs_{sex}_sn'] = arm_passes(row.interaction, lake_h[g] - sn[g]) if human_testable and sn_testable else None
        all_tests[f'mouse_sn_{sex}'] = arm_passes(row.mouse, sn[g]) if sn_testable else None
    for sex, tpm in (('male', male_tpm), ('female', female_tpm)):
        peak = max_or_nan(tpm.loc[g])
        all_tests[f'mouse_micro_{sex}'] = arm_passes(row.mouse, micro_contrast[sex][g]) if peak >= TPM_FLOOR else None
    tests = {k: all_tests[k] for k in claimed_tests(row)}
    probe = probe_table.set_index('mouse_symbol').loc[g]
    balanced = bool(probe.mouse_probes == probe.human_probes)
    detection_ok = bool(detection.loc[g].min() >= .10) or one != ''
    verdict = gene_verdict(tests, critical={'ours', 'human_lake', 'mouse_absent', 'human_absent'},
                           probe_balanced=balanced, detection_ok=detection_ok)
    if row.role == 'example':
        verdict = 'example (species claim dropped a priori)'
    evidence_rows.append({
        'story': row.story, 'gene': g, 'role': row.role, 'claim_mouse': row.mouse, 'claim_human': row.human,
        'claim_interaction': row.interaction, 'one_species': one,
        'ours_mouse': ours_m[g], 'ours_human': ours_h[g], 'ours_interaction': ours_int[g],
        'lake_human': lake_h[g], 'lake_human_S2_minus_S1': lake_h_s2[g], 'lake_human_sensitivity': lake_h_sens[g],
        'lake_human_male': lake_h_male[g], 'lake_human_female': lake_h_female[g], 'benchmark_human': bench_h[g],
        'mouse_sn_male': sn_m[g], 'mouse_sn_female': sn_f[g],
        'mouse_micro_male': micro_contrast['male'][g], 'mouse_micro_female': micro_contrast['female'][g],
        'mouse_micro_male_S2_minus_S1': micro_s2['male'][g], 'rat_micro': micro_contrast['rat'][g],
        'matched_anatomy_interaction': lake_h[g] - micro_s2['male'][g],
        'lake_max_cpm': lake_segment_cpm.loc[g].max(), 'micro_male_tpm_S1_S2_S3': '/'.join(f'{v:.1f}' for v in male_tpm.loc[g]),
        'micro_female_tpm_S1_S2_S3': '/'.join(f'{v:.1f}' for v in female_tpm.loc[g]),
        'detection_mouse': detection.loc[g, 'detection_mouse'], 'detection_human': detection.loc[g, 'detection_human'],
        'mouse_probes': int(probe.mouse_probes), 'human_probes': int(probe.human_probes), 'probe_balanced': balanced,
        'T_spatial': gene_stats.loc[g, 'T_spatial'],
        **{f'test_{k}': v for k, v in tests.items()}, 'verdict': verdict})
evidence = pd.DataFrame(evidence_rows)

conventional = pd.read_csv(upstream31 / 'conventional_deseq2_all_partitions.csv')
conventional = conventional[conventional.partition.eq('species')]
lfc = conventional.pivot(index='gene', columns='scope', values='log2FoldChange')
evidence = evidence.merge(lfc.add_prefix('deseq2_log2fc_').reset_index(), on='gene', how='left')

story_rows = []
for story in STORY_ORDER:
    block = evidence[evidence.story.eq(story)].set_index('gene')
    if story == 'probe_example':
        story_rows.append({'story': story, 'verdict': 'example', 'share_balanced_members_passing': np.nan,
                           'leads': '', 'genes': ', '.join(block.index)})
        continue
    leads = block.index[block.role.eq('lead')].tolist()
    verdict, share = story_verdict(block.verdict, leads, block.probe_balanced, can_lead=story in CAN_LEAD)
    story_rows.append({'story': story, 'verdict': verdict, 'share_balanced_members_passing': share,
                       'leads': ', '.join(f'{g} ({block.loc[g, "verdict"]})' for g in leads),
                       'genes': ', '.join(f'{g}: {v}' for g, v in block.verdict.items())})
stories = pd.DataFrame(story_rows)
stories.to_csv(output / 'story_verdicts.csv', index=False)
display(stories)
test_columns = [c for c in evidence.columns if c.startswith('test_')]
display(evidence[['story', 'gene', 'verdict', 'mouse_probes', 'human_probes'] + test_columns])
```

### Annotations added after the first run (they do not change any verdict)

The rules treat a reference below the expression floor as *not testable*. That hides one real
signal: a gene present in male but absent in female mouse PT is a sex-specific program, not a
missing measurement. Three more descriptive flags record disagreement between human references.
- `female_mouse_absent`: male microdissection peak ≥ 5 TPM and female peak < 5 TPM.
- `human_sexes_disagree`: Lake male and female S3 − S1 have opposite signs, both |·| ≥ 0.1.
- `human_references_disagree`: Lake and the benchmark human reference have opposite signs, both |·| ≥ 0.1.
- `matched_anatomy_agrees`: the claimed interaction sign also holds when the mouse arm is the
  cortical medullary-ray segment (microdissected PTS2 − PTS1) instead of S3 − S1.

```python
def opposite(a, b, floor=rules.DEAD_ZONE):
    return bool(abs(a) >= floor and abs(b) >= floor and np.sign(a) != np.sign(b))


male_peak = male_tpm.max(axis=1)
female_peak = female_tpm.max(axis=1)
evidence['female_mouse_absent'] = [bool(male_peak[g] >= TPM_FLOOR and female_peak[g] < TPM_FLOOR) for g in evidence.gene]
evidence['human_sexes_disagree'] = [opposite(a, b) for a, b in zip(evidence.lake_human_male, evidence.lake_human_female)]
evidence['human_references_disagree'] = [opposite(a, b) for a, b in zip(evidence.lake_human, evidence.benchmark_human)]
evidence['matched_anatomy_agrees'] = [arm_passes(c, v) if c else None
                                      for c, v in zip(evidence.claim_interaction, evidence.matched_anatomy_interaction)]
flag_columns = ['female_mouse_absent', 'human_sexes_disagree', 'human_references_disagree', 'matched_anatomy_agrees']
display(evidence[['story', 'gene', 'verdict'] + flag_columns])
evidence.to_csv(output / 'story_gene_evidence.csv', index=False)
display(evidence[['story', 'gene', 'verdict', 'hpa_pt_level', 'hpa_reliability']])
```

## 10 · Figure 12 · Replication matrix

One row per claimed gene, one column per pre-specified test. Filled dark square with ✓ = passes;
open square with × = fails; light grey = not testable; blank = not claimed. Verdict, probe
counts (mouse/human) and the post-hoc annotations at the right.

```python
order = evidence.assign(_s=evidence.story.map({s: i for i, s in enumerate(STORY_ORDER)})).sort_values(['_s', 'role']).reset_index(drop=True)
shown_tests = ['test_ours', 'test_ours_attenuation', 'test_human_lake', 'test_int_vs_male_sn', 'test_int_vs_female_sn',
               'test_mouse_sn_male', 'test_mouse_sn_female', 'test_mouse_micro_male', 'test_mouse_micro_female',
               'test_mouse_absent', 'test_human_absent']
labels = ['ours', 'ours (de-attenuated)', 'Lake human', 'interaction\n(male mouse sn)', 'interaction\n(female mouse sn)',
          'mouse sn ♂', 'mouse sn ♀', 'mouse micro ♂', 'mouse micro ♀', 'absent in mouse', 'absent in human']
def claim_view(r):
    return pd.Series({'mouse': r.claim_mouse, 'human': r.claim_human, 'interaction': r.claim_interaction,
                      'one_species': r.one_species})


fig, ax = plt.subplots(figsize=(11.2, .24 * len(order) + 1.6))
for i, row in order.iterrows():
    for j, key in enumerate(shown_tests):
        value = row[key]
        if key.replace('test_', '') not in claimed_tests(claim_view(row)):
            continue
        if value is None or (isinstance(value, float) and np.isnan(value)):
            ax.add_patch(plt.Rectangle((j - .4, i - .4), .8, .8, color='#E3E3E3', lw=0))
        elif bool(value):
            ax.add_patch(plt.Rectangle((j - .4, i - .4), .8, .8, color=INK, lw=0))
            ax.text(j, i, '✓', ha='center', va='center', color='white', fontsize=7)
        else:
            ax.add_patch(plt.Rectangle((j - .4, i - .4), .8, .8, facecolor='white', edgecolor=INK, lw=.8))
            ax.text(j, i, '×', ha='center', va='center', color=INK, fontsize=8)
    notes = [label for flag, label in (('female_mouse_absent', 'absent in female mouse'),
                                        ('human_sexes_disagree', 'human sexes disagree'),
                                        ('human_references_disagree', 'human references disagree')) if row[flag]]
    if row.matched_anatomy_agrees is False:
        notes.append('flips vs mouse PTS2')
    text = f'{row.verdict}  ·  M{row.mouse_probes}/H{row.human_probes}' + (f'  ·  {"; ".join(notes)}' if notes else '')
    ax.text(len(shown_tests) - .3, i, text, va='center', fontsize=7,
            color=INK if row.verdict in ('headline', 'supporting') else MUTED)
ax.set_xlim(-.6, len(shown_tests) + 5.5)
ax.set_ylim(len(order) - .5, -.5)
ax.set_yticks(range(len(order)), [f'{r.story} · {r.gene}' for r in order.itertuples()], fontsize=7)
ax.set_xticks(range(len(shown_tests)), labels, rotation=55, ha='right', fontsize=7)
ax.tick_params(length=0)
for side in ('left', 'bottom'):
    ax.spines[side].set_visible(False)
boundaries = np.flatnonzero(order.story.to_numpy()[1:] != order.story.to_numpy()[:-1]) + .5
for b in boundaries:
    ax.axhline(b, color='#CCCCCC', lw=.6)
save(fig, 'fig12_r5_replication_matrix')
plt.show()
plt.close(fig)
```
